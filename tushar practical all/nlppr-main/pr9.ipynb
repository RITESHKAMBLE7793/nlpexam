```python
# Agglomerative Clustering on Wine Quality Dataset

import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import accuracy_score

# Load dataset (download winequality-red.csv from UCI)
df = pd.read_csv("winequality-red.csv", sep=';')

# Features and target
X = df.drop("quality", axis=1)
y = df["quality"]

# Convert quality into 3 classes (low, medium, high)
y = y.apply(lambda x: 0 if x <= 4 else (1 if x <= 6 else 2))

# Normalize data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply Agglomerative Clustering
model = AgglomerativeClustering(n_clusters=3)
clusters = model.fit_predict(X_scaled)

# Evaluate (simple matching)
print("Sample Predictions:", clusters[:10])

# Optional: Compare with actual (rough idea)
print("Note: Clustering is unsupervised, so accuracy may not be exact.")
```
