```python
# Simple Co-reference Resolution

import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag

# Download once
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')

text = "Ravi has a dog. He loves it."

# Tokenize & POS tag
words = word_tokenize(text)
tags = pos_tag(words)

# Simple resolution logic
last_noun = None
resolved = []

for word, tag in tags:
    # If noun, store it
    if tag.startswith('NN'):
        last_noun = word
        resolved.append(word)
    
    # If pronoun, replace with last noun
    elif tag in ['PRP', 'PRP$']:
        if last_noun:
            resolved.append(last_noun)
        else:
            resolved.append(word)
    else:
        resolved.append(word)

print("Original:", text)
print("Resolved:", " ".join(resolved))
```
